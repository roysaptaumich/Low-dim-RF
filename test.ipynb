```python
class sigmoid:
    def __init__(self, x, scale=2):
        self.x = x
        self.scale  = scale

    def scale_x(self):
        return self.x * self.scale
    
    def apply_sigmoid(self):
        u = self.scale_x()
        return u/(1+u)
        
    
sigmoid = sigmoid(x = 3, scale  = 1)
print(sigmoid.scale_x())
```
